```python
df["daily price change"] = df["close"] - df["open"]
df["daily price change"]
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)
plt.title("Trading volume vs daily price change")
plt.xlabel("volume")
plt.ylabel("Daily Price Change")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)
plt.title("Volume vs daily price change")
plt.xlabel("volume")
plt.ylabel("Daily Price Change")

plt.show()
```

```python
#calculate the 20 days\
df["SMA_20"]= df['close'].rolling(window=20).mean()
df["SMA_20"]
```

```python
#Plot Closing Pricing + 20 Day SMA
plt.figure(figsize=(14,6))
sns.lineplot(
    data=df,
    x=df.index,
    y="close",
    label="Closing Price")

sns.lineplot(
    data=df,
    x=df.index,
    y="SMA_20",
    label="20-Day SMA")

plt.title("Shopify Closing Price and 20-Day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#daily return %
df["daily return %"] = ((df["close"]) - (df["open"])) / (df["open"]) * 100
```

```python
df["daily return %"].head()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="daily return %",
    bins=20,
    kde=True
)

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return %")
plt.ylabel("Frequency")

plt.show()
```
